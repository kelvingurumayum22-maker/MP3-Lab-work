```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-2, 2, 500)
y = x**2
z = x + 1j * y
w = z**2
plt.figure(figsize=(8, 6))
plt.plot(np.real(z), np.real(w),
         label='Real[f(z)] vs Real[z]', color='blue')
plt.plot(np.imag(z), np.imag(w),
         label='Imag[f(z)] vs Imag[z]', color='red')
plt.xlabel('Components of z')
plt.ylabel('Components of f(z)')
plt.title('Complex Function Mapping: f(z) vs z')
plt.legend()
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt
def f(x):
    return x**2
x = np.linspace(0, 10, 100)
h = x[1] - x[0]
y = f(x)
dy_forward = (f(x + h) - f(x)) / h
dy_backward = (f(x) - f(x - h)) / h
dy_central = (f(x + h) - f(x - h)) / (2 * h)
dy_exact = 2 * x
plt.figure(figsize=(9, 6))
plt.plot(x, dy_exact, 'k-', label='Exact derivative')
plt.plot(x, dy_forward, 'r--', label='Forward difference')
plt.plot(x, dy_backward, 'b--', label='Backward difference')
plt.plot(x, dy_central, 'g:', label='Central difference')
plt.xlabel('x')
plt.ylabel("Derivative f(x)")
plt.title('Finite Difference Method')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5, 6, 7, 8]
N = len(x)
X = []
for k in range(N):
    s = 0
    for n in range(N):
        s += x[n] * np.exp(-2j * np.pi * k * n / N)
    X.append(s)
magnitude = np.abs(X)
plt.plot(range(N), magnitude, 'bo-')
plt.xlabel("Frequency Index")
plt.ylabel("Magnitude")
plt.title("Discrete Fourier Transform")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
v = np.array([2, 1])
theta = np.radians(45)
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
v_new = R @ v
plt.plot([0, v[0]], [0, v[1]], 'bo-', label='Original Vector')
plt.plot([0, v_new[0]], [0, v_new[1]], 'ro-', label='Rotated Vector')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('2D Vector Rotation')
plt.grid()
plt.legend()
plt.axis('equal')
plt.show()
print("Original vector:", v)
print("Rotated vector:", v_new)
```

```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
    return x**2
x = np.array([-0.77459667, 0, 0.77459667])
w = np.array([0.55555556, 0.88888889, 0.55555556])
I = np.sum(w * f(x))
print("Integral =", I)
t = np.linspace(-1, 1, 100)
plt.plot(t, f(t))
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gauss-Legendre Quadrature")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
    return x**2
x = np.array([0.415775, 2.294280, 6.289945])
w = np.array([0.711093, 0.278518, 0.010389])
I = np.sum(w * f(x))
print("Integral =", I)
t = np.linspace(0, 8, 100)
plt.plot(t, np.exp(-t) * f(t))
plt.xlabel("x")
plt.ylabel("e^(-x) f(x)")
plt.title("Gauss-Laguerre Quadrature")
plt.grid()
plt.show()
```
